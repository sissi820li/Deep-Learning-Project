### Part (c) [4 pt]

load and parse the data into two lists: sequences and labels. Create character-level stoi and itos dictionaries. Reserve the index 0 for padding. Convert the sequences to list of character ids using stoi dictionary and convert the labels to a list of 0s and 1s by assinging class "ham" to 0 and class "spam" to 1.

````python
sequences = []
labels = []

for line in open('SMSSpamCollection'):
    if line[0] == 'h':
        labels.append('ham')
        sequences.append(line[3:].strip())
    elif line[0] == 's':
        labels.append('spam')
        sequences.append(line[4:].strip())

unique_chars = set(''.join(sequences)) #join all sequences into a single string and remove duplicate characters
stoi = {'<PAD>': 0}
itos = {0: '<PAD>'}

#Map each unique character to a unique integer ID, starting from 1 (0 is reserved for padding)
for idx, char in enumerate(sorted(unique_chars), start=1):
    stoi[char] = idx
    itos[idx] = char
    #stoi[' '] = 1, stoi['a'] = 2

# Use stoi mapping to convert each char in each sequence
encoded_sequences = [[stoi[char] for char in seq] for seq in sequences]
encoded_labels = [1 if label == 'spam' else 0 for label in labels]
    
print(f"Total unique characters including padding: {len(stoi)}")
print(f"Original data: '{sequences[2]}'")
print(f"Converted sequence: {encoded_sequences[2]}")
print(f"Encoded label: spam -> {encoded_labels[2]}")
````

### Part (d) [4 pt]

Use train_test_split function from sklearn (https://scikit-learn.org/dev/modules/generated/sklearn.model_selection.train_test_split.html) to split the data indices into `train`, `valid`, and `test`. Use a 60-20-20 split. 

You saw in part (b) that there are many more non-spam messages than spam messages. This **imbalance** in our training data will be problematic for training. We can fix this disparity by duplicating spam messages in the training set, so that the training set is roughly balanced.

````python
x = encoded_sequences
y = encoded_labels

indices = list(range(len(x)))

# Separate 20% for testing and 80% for training and validation
train_val_idx, test_index = train_test_split(indices, test_size=0.20, random_state=42)

# Separate 20% of original data for validation, so take 1/4th of train_val_idx for validation
train_index, val_index = train_test_split(train_val_idx, test_size=0.25, random_state=42)

# Use indices to grab data for corresponding data sets
train_x = [x[idx] for idx in train_index]
train_y = [y[idx] for idx in train_index]

val_x = [x[idx] for idx in val_index]
val_y = [y[idx] for idx in val_index]

test_x = [x[idx] for idx in test_index]
test_y = [y[idx] for idx in test_index]

train_spam = []
for idx, item in enumerate(train_x):
    if train_y[idx] == 1:
        train_spam.append(item)
        
# duplicate each spam message 6 more times
train_x = train_x + train_spam * 6
train_y = train_y + [1] * (len(train_spam) * 6)

datasets = {
    "Training Set": train_y,
    "Validation Set": val_y,
    "Test Set": test_y
}
````

### Part (e) [4 pt]

Since each sequence has a different length, we cannot use the default DataLoader. We need to change the DataLoader such that it can pad differnt sequence sizes within the batch. To do this, we need to introduce a **collate_fn** to the DataLoader such that it uses **pad_sequence** function (https://pytorch.org/docs/stable/generated/torch.nn.utils.rnn.pad_sequence.html) to pad the sequences within the batch to the same size. 

We also need a custom Dataset class to return a pair of sequence and label for each example. Complete the code below to address these.

Hint: 
- https://stanford.edu/~shervine/blog/pytorch-how-to-generate-data-parallel
- https://plainenglish.io/blog/understanding-collate-fn-in-pytorch-f9d1742647d3

````python
class MyDataset(Dataset):
    def __init__(self, sequences, labels):
        self.sequences = sequences
        self.labels = labels
        
    def __len__(self):
        return len(self.sequences)
    
    def __getitem__(self, idx):
        # Return a single (sequence, label) pair as PyTorch tensors
        seq_tensor = torch.tensor(self.sequences[idx], dtype=torch.long)
        label_tensor = torch.tensor(self.labels[idx], dtype=torch.float32)
        return seq_tensor, label_tensor

def collate_sequences(batch):
    # Unzip the batch into lists
    sequences = [item[0] for item in batch]
    labels = [item[1] for item in batch]
    padded_sequences = pad_sequence(sequences, batch_first=True, padding_value=0)
    labels_tensor = torch.stack(labels) #stack labels into a single tensor
    
    return padded_sequences, labels_tensor

train_loader = DataLoader(dataset=MyDataset(train_x, train_y), batch_size=32, shuffle=True, collate_fn=collate_sequences)
val_loader = DataLoader(dataset=MyDataset(val_x, val_y), batch_size=32, shuffle=False, collate_fn=collate_sequences)
test_loader = DataLoader(dataset=MyDataset(test_x, test_y), batch_size=32, shuffle=False, collate_fn=collate_sequences)
````

Build a recurrent neural network model, using an architecture of your choosing.

recurrent network output.

Instead of using the RNN output value for the final token, another often used
strategy is to max-pool over the entire output array. That is, instead of calling
something like:

```
out, _ = self.rnn(x)
self.fc(out[:, -1, :])
```

where `self.rnn` is an `nn.RNN`, `nn.GRU`, or `nn.LSTM` module, and `self.fc` is a 
fully-connected 
layer, we use:

```
out, _ = self.rnn(x)
self.fc(torch.max(out, dim=1)[0])
```

This works reasonably in practice. An even better alternative is to concatenate the
max-pooling and average-pooling of the RNN outputs:

```
out, _ = self.rnn(x)
out = torch.cat([torch.max(out, dim=1)[0], 
                 torch.mean(out, dim=1)], dim=1)
self.fc(out)
```

We encourage you to try out all these options. The way you pool the RNN outputs
is one of the "hyperparameters" that you can choose to tune later on.

````python
# You might find this code helpful for obtaining
# PyTorch one-hot vectors.

ident = torch.eye(10)
print(ident[0]) # one-hot vector
print(ident[1]) # one-hot vector
x = torch.tensor([[1, 2], [3, 4]])
print(ident[x]) # one-hot vectors
````

````python
class RNN(nn.Module):
  def __init__(self, vocab_size, hidden_size, n_layers=1):
      super(RNN, self).__init__()
      
      self.ident = torch.eye(vocab_size)
      self.rnn = nn.GRU(vocab_size, hidden_size, n_layers, batch_first=True)
      self.classifier = nn.Linear(hidden_size, 2)
      
  def forward(self, x):
      inp_x = self.ident[x]
      output, _ = self.rnn(inp_x)

      out = torch.max(output, dim=1)[0]
      
      out = self.classifier(out)
      return out
````

## Part 3. Training [16 pt]

### Part (a) [4 pt]

Complete the `get_accuracy` function, which will compute the
accuracy (rate) of your model across a dataset (e.g. validation set).

````python
def get_accuracy(model, data_loader):
    """ Compute the accuracy of the `model` across a dataset `data`
    
    Example usage:
    
    >>> model = MyRNN() # to be defined
    >>> get_accuracy(model, valid) # the variable `valid` is from above
    """
    correct = 0
    total = 0

    with torch.no_grad():  # Turn off gradient tracking to save memory and time
        for sequences, labels in data_loader:
            output = model(sequences)
            pred = output.max(1, keepdim=True)[1] # Get pred (0 or 1)
            correct += pred.eq(labels.view_as(pred)).sum().item() # Compare predictions to labels
            total += labels.shape[0]
        
    return correct / total

def evaluate(model, data_loader, criterion):
    """ Evaluate the network on the validation/test set. """
    total_loss = 0.0
    total_err = 0.0
    total_epoch = 0
    
    # FIX 3: Set model to evaluation mode and turn off gradients
    model.eval()
    with torch.no_grad():
        
        for sequences, labels in data_loader:
            labels = labels.long() 
            outputs = model(sequences)
            loss = criterion(outputs, labels)
            pred = outputs.max(1, keepdim=True)[1]
            corr = pred.ne(labels.view_as(pred))
            
            total_err += int(corr.sum().item())
            total_loss += loss.item()
            total_epoch += len(labels)
            
    err = float(total_err) / total_epoch
    avg_loss = float(total_loss) / len(data_loader)
    
    return err, avg_loss
````

### Part (b) [4 pt]

Train your model. Plot the training curve of your final model. 
Your training curve should have the training/validation loss and
accuracy plotted periodically.

Note: Not all of your batches will have the same batch size.
In particular, if your training set does not divide evenly by
your batch size, there will be a batch that is smaller than
the rest. 

````python
def get_model_name(name, batch_size, learning_rate, epoch):
    path = "model_{0}_bs{1}_lr{2}_epoch{3}".format(name,
                                                   batch_size,
                                                   learning_rate,
                                                  epoch)
    return path

def train(model, train_loader, valid_loader, num_epochs=5, learning_rate=1e-4):
    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
    
    train_err, train_loss, val_err, val_loss = [], [], [], []
    
    for epoch in range(num_epochs):
        # 1. Train the model
        model.train() # Set the model to training mode
        for sequences, labels in train_loader:
            optimizer.zero_grad()
            pred = model(sequences)
            # Ensure labels are LongTensors for CrossEntropyLoss
            loss = criterion(pred, labels.long()) 
            loss.backward()
            optimizer.step()
            
        # 2. Evaluate the model using the function from Part 3 (a)
        t_err, t_loss = evaluate(model, train_loader, criterion)
        v_err, v_loss = evaluate(model, valid_loader, criterion)
        
        train_err.append(t_err)
        train_loss.append(t_loss)
        val_err.append(v_err)
        val_loss.append(v_loss)
        
        print(("Epoch {}: Train err: {}, Train loss: {} |"+
               "Validation err: {}, Validation loss: {}").format(
                   epoch + 1,
                   t_loss,
                   v_loss,
                   t_err,
                   v_err))
        
        # 3. Save the model state at the end of each epoch
        # Default to "rnn" if the model doesn't explicitly have a name attribute
        model_name = getattr(model, 'name', 'rnn') 
        batch_size = train_loader.batch_size
        model_path = get_model_name(model_name, batch_size, learning_rate, epoch)
        torch.save(model.state_dict(), model_path)
        
    # 4. Save the performance arrays to CSV files for plot_training_curve
    # This uses the final epoch's model_path name to write the files
    np.savetxt("{}_train_err.csv".format(model_path), train_err)
    np.savetxt("{}_train_loss.csv".format(model_path), train_loss)
    np.savetxt("{}_val_err.csv".format(model_path), val_err)
    np.savetxt("{}_val_loss.csv".format(model_path), val_loss)
    
def plot_training_curve(path):
    import matplotlib.pyplot as plt
    train_err = np.loadtxt("{}_train_err.csv".format(path))
    val_err = np.loadtxt("{}_val_err.csv".format(path))
    train_loss = np.loadtxt("{}_train_loss.csv".format(path))
    val_loss = np.loadtxt("{}_val_loss.csv".format(path))
    plt.title("Train vs Validation Error")
    n = len(train_err) # number of epochs
    plt.plot(range(1,n+1), train_err, label="Train")
    plt.plot(range(1,n+1), val_err, label="Validation")
    plt.xlabel("Epoch")
    plt.ylabel("Error")
    plt.legend(loc='best')
    plt.show()
    plt.title("Train vs Validation Loss")
    plt.plot(range(1,n+1), train_loss, label="Train")
    plt.plot(range(1,n+1), val_loss, label="Validation")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend(loc='best')
    plt.show()
````

````python
vocab_size = len(stoi) # Number of unique characters including padding
model = RNN(vocab_size=vocab_size, hidden_size=50)
train(model, train_loader, val_loader, learning_rate=0.001, num_epochs = 30)
model_path = get_model_name("rnn", 32, 0.001, 29)
plot_training_curve(model_path)
````
